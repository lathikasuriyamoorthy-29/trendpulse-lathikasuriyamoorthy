```python
from google.colab import files

uploaded = files.upload()
```

```python
import pandas as pd
import numpy as np
```

```python
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Loaded data:", df.shape)
```

```python
print("\nFirst 5 rows:")
print(df.head())
```

```python
average_score = np.mean(df["score"])

print("Average score:", average_score)
```

```python
average_comments = np.mean(df["num_comments"])

print("Average comments:", average_comments)
```

```python
df["score"].describe()
```

```python
mean_score = np.mean(df["score"])
print("\n mean for score:\n",mean_score)
median_score = np.median(df["score"])
print("\n median for score:\n",median_score)
std_score = np.std(df["score"])
print("\n standard deviation of score:\n",std_score)
max_score = np.max(df["score"])
print("\n maximum of score:\n",max_score)
min_score = np.min(df["score"])
print("\n minimum of score:\n",min_score)
```

```python
category_counts = df["category"].value_counts()
```

```python
most_common_category = category_counts.idxmax()
most_common_count = category_counts.max()
print(
    f"Most stories in: {most_common_category} "
    f"({most_common_count} stories)"
)
```

```python
most_commented = df.loc[df["num_comments"].idxmax()]
print(
    f'Most commented story: "{most_commented["title"]}" '
    f'— {most_commented["num_comments"]} comments')
```

```python
df["engagement"] = df["num_comments"]/(df["score"]+1)
```

```python
df["is_popuar"] = df["engagement"] > mean_score
```

```python
output_file = "trends_analyzed.csv"

df.to_csv(output_file, index=False)

print(f"\nSaved to {output_file}")
```
